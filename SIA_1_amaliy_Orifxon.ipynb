```python
import pandas as pd

data = {
    "name": ["Ali", "Vali", "Hasan", "Husan", "Karim"],
    "age": [25, 32, 28, 45, 36],
    "salary": [300, 450, 380, 600, 520],
    "department": ["IT", "HR", "IT", "Finance", "HR"]
}

df = pd.DataFrame(data)
df
```

```python
def remove_outliers_iqr(dataframe, column):
    Q1 = dataframe[column].quantile(0.25)
    Q3 = dataframe[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    return dataframe[
        (dataframe[column] >= lower) &
        (dataframe[column] <= upper)
    ]

df_clean = remove_outliers_iqr(df, "salary")
df_clean
```
